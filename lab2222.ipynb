task 1:

```python
import pandas as pd
import re

data = pd.read_csv('/content/apple-twitter-sentiment-texts.csv')

hashtags = []

for text in data['text']:
    hashtags.extend(re.findall(r"#\w+", text))

print(len(hashtags))
hashtags = [hashtag.lower() for hashtag in hashtags]

print(pd.Series(hashtags).value_counts().head(10))
```

task 2:

```python
import re

text = "This year is 2021"

pattern = re.compile(r"\d+")

print(type(pattern))

text = pattern.sub("2022", text)

print(text)
```

The re.compile() function builds a pattern object. It comes in handy in cases where we need to apply the same pattern several times.

```python
import re

text = "a 11 b 2 3 c 4"

print(re.split(r"\d+", text))
```

re.split() will split the text every time the pattern occurs.

spaCy:


```python
import spacy

nlp = spacy.load('en_core_web_sm')

text = "I'm enjoying the NLP course!"

for token in nlp(text):
    print(token)
```

The language model used to analyze the text is loaded by spacy.load().

NLTK:

```python
import nltk

text = "I'm enjoying the NLP course!"

print(nltk.word_tokenize(text))
```

spaCy and NLTK both provide identical tokens for the above sentence. Both tokenize "I'm" into two tokens and separate the exclamation mark.
